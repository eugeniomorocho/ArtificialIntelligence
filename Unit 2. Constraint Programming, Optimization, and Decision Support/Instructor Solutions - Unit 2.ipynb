Keep this answer key separate from the student notebook. One valid schedule is shown below; search order can produce a different valid result.

```python
students = ["Ada", "Ben", "Cy", "Dee"]
slots = ["9:00", "10:00", "11:00"]
domains = {student: slots[:] for student in students}
conflicts = {
    "Ada": ["Ben"],
    "Ben": ["Ada", "Cy"],
    "Cy": ["Ben", "Dee"],
    "Dee": ["Cy"],
}

def is_consistent(student, slot, assignment):
    return all(assignment.get(partner) != slot for partner in conflicts[student])

def search(assignment):
    if len(assignment) == len(students):
        return assignment.copy()
    student = next(s for s in students if s not in assignment)
    for slot in domains[student]:
        if is_consistent(student, slot, assignment):
            assignment[student] = slot
            result = search(assignment)
            if result is not None:
                return result
            del assignment[student]
    return None

schedule = search({})
print(schedule)
assert schedule is not None
assert all(schedule[a] != schedule[b] for a, partners in conflicts.items() for b in partners)
```

A possible output is `{'Ada': '9:00', 'Ben': '10:00', 'Cy': '9:00', 'Dee': '10:00'}`. The consistency test prunes a branch as soon as a student conflicts with an already scheduled partner. For an added conflict, update both students' neighbor lists; the reciprocal representation keeps the model easy to inspect.
