Reference implementation for the student exercise. The model assumes the observed features are conditionally independent given the class.

```python
def posterior_spam(prior, likelihoods_spam, likelihoods_ham):
    spam_weight = prior
    ham_weight = 1 - prior
    for likelihood in likelihoods_spam:
        spam_weight *= likelihood
    for likelihood in likelihoods_ham:
        ham_weight *= likelihood
    return spam_weight / (spam_weight + ham_weight)

p_both = posterior_spam(0.20, [0.70, 0.80], [0.10, 0.20])
p_offer_only = posterior_spam(0.20, [0.70, 0.20], [0.10, 0.80])
threshold = 4 / (4 + 1)
print(f"Both features: {p_both:.1%}; offer only: {p_offer_only:.1%}; threshold: {threshold:.1%}")
print("Decisions:", p_both > threshold, p_offer_only > threshold)
```

The posteriors are 87.5% with both features and about 30.4% with an offer but no link. At an 80% threshold, only the first message is classified as spam. Conditional independence may fail because features such as promotional wording and links are often generated together by the same sender or campaign.
