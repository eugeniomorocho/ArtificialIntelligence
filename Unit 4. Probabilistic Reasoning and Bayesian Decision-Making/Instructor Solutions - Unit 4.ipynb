# Instructor reference — Unit 4 weather-decision example

Worked calculations for the illustrative umbrella example. The prior and app alert rates are teaching assumptions, not a real forecast.

```python
def posterior(prior, likelihoods_if_rain, likelihoods_if_dry):
    rain_weight = prior
    dry_weight = 1 - prior
    for likelihood in likelihoods_if_rain:
        rain_weight *= likelihood
    for likelihood in likelihoods_if_dry:
        dry_weight *= likelihood
    return rain_weight / (rain_weight + dry_weight)

def choose_umbrella(p_rain, umbrella_dry_cost=2, soaked_cost=4):
    expected_cost_bring = (1 - p_rain) * umbrella_dry_cost
    expected_cost_leave = p_rain * soaked_cost
    return expected_cost_bring < expected_cost_leave

prior_rain = 0.30
p_alert_given_rain, p_alert_given_dry = 0.80, 0.25

p_alert = posterior(prior_rain, [p_alert_given_rain], [p_alert_given_dry])
p_alert_rare_rain = posterior(0.10, [p_alert_given_rain], [p_alert_given_dry])
p_alert_common_rain = posterior(0.60, [p_alert_given_rain], [p_alert_given_dry])
threshold = 2 / (2 + 4)

print(f"Prior rain probability: {prior_rain:.1%}; bring umbrella: {choose_umbrella(prior_rain)}")
print(f"Rain given alert (10% prior): {p_alert_rare_rain:.1%}")
print(f"Rain given alert (30% prior): {p_alert:.1%}")
print(f"Rain given alert (60% prior): {p_alert_common_rain:.1%}")
print(f"Umbrella decision threshold: {threshold:.1%}")
print("Bring umbrella after alert:", choose_umbrella(p_alert_rare_rain),
      choose_umbrella(p_alert), choose_umbrella(p_alert_common_rain))

assert abs(p_alert - 0.24 / 0.415) < 1e-12
assert abs(p_alert_rare_rain - 0.08 / 0.305) < 1e-12
assert abs(p_alert_common_rain - 0.48 / 0.58) < 1e-12
assert choose_umbrella(prior_rain) is False
assert choose_umbrella(p_alert_rare_rain) is False
assert choose_umbrella(p_alert) is True
assert choose_umbrella(p_alert_common_rain) is True
```

With a 30% prior, 240 of the 415 alert mornings are rainy, so the posterior is about 57.8%. When the prior is 10%, the posterior after an alert is about 26.2%; when the prior is 60%, it is about 82.8%. The one-third cost threshold means not bringing an umbrella in the first case, but bringing one for 30% and 60% priors after an alert.
