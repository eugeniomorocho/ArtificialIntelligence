# Instructor reference — Unit 4 AI-detector example

Worked calculations for the illustrative AI-writing-detector example. The prior, the detector rates, and the costs are teaching assumptions, not measurements of any real detector.

```python
def posterior(prior, likelihoods_if_ai, likelihoods_if_human):
    ai_weight = prior
    human_weight = 1 - prior
    for likelihood in likelihoods_if_ai:
        ai_weight *= likelihood
    for likelihood in likelihoods_if_human:
        human_weight *= likelihood
    return ai_weight / (ai_weight + human_weight)

def choose_penalize(p_ai, wrongful_cost=9, missed_cost=1):
    expected_cost_penalize = (1 - p_ai) * wrongful_cost
    expected_cost_not_penalize = p_ai * missed_cost
    return expected_cost_penalize < expected_cost_not_penalize

flag_if_ai, flag_if_human = 0.90, 0.05

p_rare = posterior(0.02, [flag_if_ai], [flag_if_human])
p_base = posterior(0.10, [flag_if_ai], [flag_if_human])
p_common = posterior(0.50, [flag_if_ai], [flag_if_human])
threshold = 9 / (9 + 1)

print(f"AI given flag (2% prior): {p_rare:.1%}")
print(f"AI given flag (10% prior): {p_base:.1%}")
print(f"AI given flag (50% prior): {p_common:.1%}")
print(f"Penalize threshold: {threshold:.1%}")
print("Penalize on flag alone:", choose_penalize(p_rare),
      choose_penalize(p_base), choose_penalize(p_common))

assert abs(p_rare - 0.018 / 0.067) < 1e-12
assert abs(p_base - 0.09 / 0.135) < 1e-12
assert abs(p_common - 0.45 / 0.475) < 1e-12
assert choose_penalize(p_rare) is False
assert choose_penalize(p_base) is False
assert choose_penalize(p_common) is True
```

With a 10% prior, 90 of the 135 flagged essays are AI-written, so the posterior is about 66.7%. With a 2% prior it is about 26.9%; with a 50% prior it is about 94.7%. The 90% cost threshold means a flag alone justifies a penalty only in the 50% case.
