# Instructor reference — Unit 4 weather example

Worked calculations for the clouds-and-rain umbrella example, based on the CS50 AI (Lecture 2) weather numbers. The rates and costs are teaching assumptions, not a real forecast.

```python
def bayes(prior, likelihood, evidence_rate):
    return likelihood * prior / evidence_rate

def choose_umbrella(p_rain, dry_cost=2, soaked_cost=4):
    return (1 - p_rain) * dry_cost < p_rain * soaked_cost

cloudy_rate, cloudy_given_rain = 0.40, 0.80

p_5 = bayes(0.05, cloudy_given_rain, cloudy_rate)
p_10 = bayes(0.10, cloudy_given_rain, cloudy_rate)
p_25 = bayes(0.25, cloudy_given_rain, cloudy_rate)
p_clear = (1 - cloudy_given_rain) * 0.10 / (1 - cloudy_rate)
threshold = 2 / (2 + 4)

print(f"Rain given cloudy (5%, 10%, 25% priors): {p_5:.1%}, {p_10:.1%}, {p_25:.1%}")
print(f"Rain given clear morning (10% prior): {p_clear:.1%}")
print(f"Umbrella threshold: {threshold:.1%}")
print("Bring umbrella:", choose_umbrella(p_5), choose_umbrella(p_10), choose_umbrella(p_25))

assert abs(p_5 - 0.10) < 1e-12
assert abs(p_10 - 0.20) < 1e-12
assert abs(p_25 - 0.50) < 1e-12
assert abs(p_clear - 20 / 600) < 1e-12
assert choose_umbrella(p_5) is False
assert choose_umbrella(p_10) is False
assert choose_umbrella(p_25) is True
```

With a 10% rain rate, 80 of the 400 cloudy mornings are followed by rain, so the posterior is 20% (clouds double the chance but it stays below the one-third threshold). With 5% it is 10%; with 25% it is 50%. Only the 25% case justifies bringing the umbrella. A clear morning gives about 3.3% (20 of 600).
