# **Bayes' Theorem: What Does "Flagged" Really Tell Us?**

**Setup:** Python 3; no external packages required.

Imagine you write an essay entirely by yourself and submit it. The course's AI-writing detector marks it as **"AI-generated."** The detector's website says it "catches 90% of AI-written essays." Does that mean there is a 90% chance that you cheated?

Most people answer "yes." By the end of this notebook you will see that the true chance is much lower, and why that matters when a flag can damage someone's record.

To find out, we will take the seat of the academic-integrity reviewer who receives the flagged essay and must decide what to do. The question is:

**Given that an essay was flagged, how likely is it really AI-written?**

**Learning goals:** answer that question from the evidence, explain every probability in the calculation, and choose an action by comparing the cost of each possible mistake.

## 1. What we know, what we cannot know, and why

Before calculating, we should ask where our information comes from. This is the key to the whole lesson.

**What we have:**

- **From the vendor's lab test.** The vendor tested the detector on essays whose origin was known for certain: essays they generated with AI, and essays written by humans before AI writing tools existed. The detector flagged **90%** of the AI-written essays (its *hit rate*) and also flagged **5%** of the human-written essays (its *false-alarm rate*).
- **From an anonymous survey of our course.** About **10%** of the essays submitted here are AI-written. This is how common the problem is *before* looking at any detector result.

**What we do not have, and cannot get:** a table of this course's real essays, each labeled "AI-written" or "human-written." Nobody knows which submitted essays are truly AI-written. That is exactly why we bought a detector. If we had those labels, we could simply count and we would not need Bayes' theorem.

Notice also that the vendor's lab groups were chosen by the vendor (for example, equal numbers of AI and human essays), so their counts say nothing about how common AI essays are in our course. The detector's rates and the survey come from different sources, and Bayes' theorem is the tool that combines them.

**The bookkeeping trick.** To combine them, imagine 1,000 essays submitted to this course. These are imagined counts, used only to keep track of proportions:

- 10% are AI-written: $0.10\times1000=100$ AI essays, and $900$ human essays.
- The detector flags 90% of the 100 AI essays: $0.90\times100=90$ **correct flags**.
- The detector flags 5% of the 900 human essays: $0.05\times900=45$ **false flags**.

That makes $90+45=135$ flagged essays, and only 90 of them are really AI-written. So among flagged essays, the fraction that are AI-written is $90/135\approx66.7\%$. About **one in three flagged essays belongs to an honest student**, even though the detector "catches 90%."

### The imagined essays at a glance

$$
\begin{array}{lrrr}
\hline
\textbf{Essay origin} & \textbf{Essays} & \textbf{Flagged} & \textbf{Not flagged} \\
\hline
\text{AI-written} & 100 & 90 & 10 \\
\text{Human-written} & 900 & 45 & 855 \\
\hline
\textbf{Total} & \textbf{1{,}000} & \textbf{135} & \textbf{865} \\
\hline
\end{array}
$$

Look only at the **Flagged** column: 90 of its 135 essays are AI-written. The 90% hit rate describes a different question, namely the **AI-written row**: 90 of its 100 essays are flagged.

```python
# Given numbers (vendor lab test and course survey); we imagine 1,000 essays.
imagined_essays = 1_000
ai_share = 0.10                      # from the course survey
flag_rate_on_ai = 0.90               # vendor test: hit rate
flag_rate_on_human = 0.05            # vendor test: false-alarm rate

ai_essays = imagined_essays * ai_share
human_essays = imagined_essays - ai_essays
correct_flags = ai_essays * flag_rate_on_ai
false_flags = human_essays * flag_rate_on_human
all_flags = correct_flags + false_flags

print(f"AI-written essays: {ai_essays:.0f}; human-written essays: {human_essays:.0f}")
print(f"Correct flags: {correct_flags:.0f}")
print(f"False flags (honest students): {false_flags:.0f}")
print(f"All flagged essays: {all_flags:.0f}")
print(f"Chance a flagged essay is AI-written: {correct_flags / all_flags:.1%}")
```

### Naming the pieces

To avoid writing long sentences, let $A$ mean **the essay is AI-written** and $F$ mean **the essay is flagged**. The vertical bar $\mid$ means "looking only at the cases where" or "given that." The bar is not a division sign.

- $P(A\mid F)$: among **flagged** essays, the fraction that are AI-written. This is what the reviewer needs. Here it is $90/135$.
- $P(F\mid A)$: among **AI-written** essays, the fraction that are flagged. This is the vendor's hit rate, $0.90$.
- $P(F\mid\text{human})$: among **human-written** essays, the fraction that are flagged. This is the false-alarm rate, $0.05$.
- $P(A)$: among **all** essays, the fraction that are AI-written. This is the **prior**, $0.10$: how common AI essays are before we look at any flag.

Each probability refers to a different group of essays: flagged essays, AI essays, human essays, or all essays. Mixing up $P(A\mid F)$ and $P(F\mid A)$ means mixing up two different groups.

## 2. Now translate the counting into Bayes' theorem

A probability such as $P(A\cap F)$ is the chance that **both** things happen to the same essay: it is AI-written **and** flagged. The symbol $\cap$ means "and at the same time." We use it instead of a comma so the meaning is visible.

$$
P(A\mid F)=\frac{P(A\cap F)}{P(F)},\qquad P(F)>0.
$$

Read this as: **among all flagged essays**, count those that are AI-written. The condition $P(F)>0$ says that at least some essays must be flagged; if none were, "the fraction of flagged essays that are AI-written" would have no denominator.

Bayes' theorem is not a mysterious reversal. It follows from counting the **same AI-and-flagged essays** in two ways:

1. Start with all essays and ask what fraction are both AI-written and flagged:

$$P(A\cap F)=P(F\mid A)\,P(A).$$

This says: the chance an essay is AI-written **and** flagged equals the chance it is AI-written, multiplied by the fraction of AI essays that get flagged. With our numbers: $0.90\times0.10=0.09$, which is the 90 essays out of 1,000.

2. Start instead with flagged essays and ask what fraction of all essays are both flagged and AI-written:

$$P(A\cap F)=P(A\mid F)\,P(F).$$

Both expressions equal the same quantity, so set them equal and divide by $P(F)$:

$$
P(A\mid F)=\frac{P(F\mid A)\,P(A)}{P(F)}.
$$

This is Bayes' theorem. It does not claim $P(A\mid F)=P(F\mid A)$. It gives a way to calculate the first from the second while accounting for how common AI essays are.

We still need the denominator: how common are flags overall? Flagged essays come from exactly two groups, AI-written and human-written, so add what each group contributes:

$$
P(F)=P(F\mid A)\,P(A)+P(F\mid\text{human})\,P(\text{human}).
$$

Substituting this into Bayes' theorem gives the full calculation:

$$
P(A\mid F)=\frac{P(F\mid A)\,P(A)}{P(F\mid A)\,P(A)+P(F\mid\text{human})\,P(\text{human})}.
$$

The numerator counts the **correct flags**. The denominator counts **all flags**: correct flags plus false flags. That is why the prior $P(A)$ matters: when AI essays are rare, the many honest essays produce many false flags.

```python
prior_ai = 0.10
flag_rate_on_ai = 0.90
flag_rate_on_human = 0.05

chance_of_any_flag = (
    flag_rate_on_ai * prior_ai
    + flag_rate_on_human * (1 - prior_ai)
)
chance_ai_given_flag = flag_rate_on_ai * prior_ai / chance_of_any_flag

print(f"P(F) = {chance_of_any_flag:.3f}")
print(f"P(A | F) = {chance_ai_given_flag:.1%}")
assert abs(chance_ai_given_flag - 90 / 135) < 1e-12
```

```python
prior_ai = 0.10
flag_rate_if_ai = 0.90
flag_rate_if_human = 0.05

ai_and_flagged = prior_ai * flag_rate_if_ai
human_and_flagged = (1 - prior_ai) * flag_rate_if_human
all_flagged = ai_and_flagged + human_and_flagged
posterior_ai = ai_and_flagged / all_flagged

print(f"AI-written AND flagged: {ai_and_flagged:.3f}")
print(f"Human-written AND flagged: {human_and_flagged:.3f}")
print(f"All flagged: {all_flagged:.3f}")
print(f"AI-written given flagged: {posterior_ai:.1%}")
assert abs(posterior_ai - 0.09 / 0.135) < 1e-12
```

## 3. From the updated probability to a decision

A probability says how plausible cheating is; it does not decide what the committee should do. We must also weigh what each mistake costs. These costs are **value judgments**, and here we assume:

- Penalizing an honest student is very harmful: **9 points**.
- Letting an AI-written essay pass without penalty is a smaller harm: **1 point**.

Let $p$ mean "the chance the essay is AI-written **given that it was flagged**." Then the chance it is human-written is $1-p$. The average (expected) cost is a probability-weighted average:

- If we **penalize**, we are wrong when the essay is human-written: average cost $9(1-p)$.
- If we **do not penalize**, we are wrong when the essay is AI-written: average cost $1\cdot p$.

Penalize only when its average cost is smaller. Solve the inequality one step at a time:

$$
\begin{aligned}
9(1-p)&<p \\
9&<10p \\
p&>\frac{9}{10}.
\end{aligned}
$$

So the committee should penalize on the flag alone only if it is more than 90% sure the essay is AI-written. Our calculation gave $p=90/135\approx66.7\%$, below that threshold. **A flag by itself is not enough** in this example: the committee should look for more evidence, such as the essay's draft history or a short conversation with the student.

The threshold comes from the costs; Bayes' theorem supplied the updated probability. A detector that "catches 90%" is not the same as a flag that is 90% reliable.

## Individual practice: what if AI essays are rarer or more common?

Keep the detector fixed: it flags 90% of AI essays and 5% of human essays. Change only how common AI essays are. This separates the **prior** (how common AI essays are before the flag) from the **evidence** (the flag).

```python
flag_rate_on_ai = 0.90
flag_rate_on_human = 0.05

def ai_chance_after_flag(prior_ai):
    """Calculate the chance an essay is AI-written after a flag, using Bayes' theorem."""
    chance_of_any_flag = (
        flag_rate_on_ai * prior_ai
        + flag_rate_on_human * (1 - prior_ai)
    )
    return flag_rate_on_ai * prior_ai / chance_of_any_flag

def should_penalize(chance_ai):
    """Return the lower-cost choice: wrongly penalizing costs 9, missing AI costs 1."""
    cost_of_penalizing = 9 * (1 - chance_ai)
    cost_of_not_penalizing = 1 * chance_ai
    return cost_of_penalizing < cost_of_not_penalizing

for prior in (0.02, 0.10, 0.50):
    posterior = ai_chance_after_flag(prior)
    print(
        f"Prior AI share: {prior:.0%}; "
        f"AI chance after flag: {posterior:.1%}; "
        f"penalize on flag alone: {should_penalize(posterior)}"
    )
```

```python
assert abs(ai_chance_after_flag(0.02) - (0.018 / 0.067)) < 1e-12
assert abs(ai_chance_after_flag(0.10) - (0.09 / 0.135)) < 1e-12
assert abs(ai_chance_after_flag(0.50) - (0.45 / 0.475)) < 1e-12
assert should_penalize(ai_chance_after_flag(0.02)) is False
assert should_penalize(ai_chance_after_flag(0.10)) is False
assert should_penalize(ai_chance_after_flag(0.50)) is True
print("Posterior calculations and decisions passed.")
```
